```python
import pandas as pd
import json

# --- CONFIG ---
input_file = "nodes_and_links.ods"
output_dir = "."   # can be changed if you want to save elsewhere

# --- READ SHEETS ---
nist_nodes_df = pd.read_excel(input_file, sheet_name="nist_nodes", engine="odf")
topics_nodes_df = pd.read_excel(input_file, sheet_name="topics_nodes", engine="odf")
families_nodes_df = pd.read_excel(input_file, sheet_name="families_nodes", engine="odf")
keywords_nodes_df = pd.read_excel(input_file, sheet_name="keywords_nodes", engine="odf")
topics_links_df = pd.read_excel(input_file, sheet_name="topics_links", engine="odf")
families_links_df = pd.read_excel(input_file, sheet_name="families_links", engine="odf")
keywords_links_df = pd.read_excel(input_file, sheet_name="keywords_links", engine="odf")

# --- CLEAN BASIC STRUCTURE ---
nist_nodes_df = nist_nodes_df.fillna("")
topics_nodes_df = topics_nodes_df.fillna("")
families_nodes_df  = families_nodes_df.fillna("")
keywords_nodes_df = keywords_nodes_df.fillna("")
topics_links_df = topics_links_df.fillna("")
families_links_df = families_links_df.fillna("")
keywords_links_df = keywords_links_df.fillna("")

# --- CONVERT TO DICTIONARIES ---
nist_nodes_json = nist_nodes_df.to_dict(orient="records")
topics_nodes_json = topics_nodes_df.to_dict(orient="records")
families_nodes_json = families_nodes_df.to_dict(orient="records")
keywords_nodes_json = keywords_nodes_df.to_dict(orient="records")
topics_links_json = topics_links_df.to_dict(orient="records")
families_links_json = families_links_df.to_dict(orient="records")
keywords_links_json = keywords_links_df.to_dict(orient="records")


# --- SAVE TO JSON FILES ---
with open(f"{output_dir}/nist_nodes.json", "w", encoding="utf-8") as f:
    json.dump(nist_nodes_json, f, indent=2, ensure_ascii=False)

with open(f"{output_dir}/topics_nodes.json", "w", encoding="utf-8") as f:
    json.dump(topics_nodes_json, f, indent=2, ensure_ascii=False)

with open(f"{output_dir}/families_nodes.json", "w", encoding="utf-8") as f:
    json.dump(families_nodes_json, f, indent=2, ensure_ascii=False)
    
with open(f"{output_dir}/keywords_nodes.json", "w", encoding="utf-8") as f:
    json.dump(keywords_nodes_json, f, indent=2, ensure_ascii=False)
    
with open(f"{output_dir}/topics_links.json", "w", encoding="utf-8") as f:
    json.dump(topics_links_json, f, indent=2, ensure_ascii=False)

with open(f"{output_dir}/families_links.json", "w", encoding="utf-8") as f:
    json.dump(families_links_json, f, indent=2, ensure_ascii=False)
    
with open(f"{output_dir}/keywords_links.json", "w", encoding="utf-8") as f:
    json.dump(keywords_links_json, f, indent=2, ensure_ascii=False)
    
    
print("✅ JSON files created: nist_nodes.json, topics_nodes.json, families_nodes.json, keywords_nodes.json, topics_links.json, families_links.json, keywords_links.json")

```

```python
from collections import defaultdict

# --- Load your existing objectives.json ---
with open("objectives.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# --- Group by id ---
grouped = defaultdict(list)
for entry in data:
    grouped[str(entry["id"])].extend(entry["objectives"])

# --- Convert to the correct format ---
result = [{"id": k, "objectives": v} for k, v in grouped.items()]

# --- Save back to a new JSON file ---
with open("objectives_grouped.json", "w", encoding="utf-8") as f:
    json.dump(result, f, indent=2, ensure_ascii=False)

print("✅ Grouped objectives.json written successfully.")
print(f"📦 Total unique IDs: {len(result)}")
```

```python
import os
import shutil
from pathlib import Path

def move_json_files(destination_relative="../html_workspace", overwrite=True):
    """
    Move all .json files from the script directory (source) into destination_relative
    (which is interpreted relative to the script directory). Works in interactive
    interpreters too (falls back to current working directory).
    """

    # Resolve source folder robustly:
    if "__file__" in globals():
        # When script is run as a file
        source_folder = Path(__file__).resolve().parent
    else:
        # When running in interactive session / notebook
        source_folder = Path.cwd()

    # Resolve destination relative to the source folder
    destination_folder = (source_folder / destination_relative).resolve()

    # Ensure destination exists
    destination_folder.mkdir(parents=True, exist_ok=True)

    # Find JSON files in source folder
    json_files = [p for p in source_folder.iterdir() if p.is_file() and p.suffix.lower() == ".json"]

    if not json_files:
        print(f"⚠️ No JSON files found in {source_folder}")
        return

    print(f"📦 Found {len(json_files)} JSON file(s) in {source_folder} — moving to {destination_folder}")

    for src_path in json_files:
        dest_path = destination_folder / src_path.name

        if dest_path.exists() and not overwrite:
            base = src_path.stem
            ext = src_path.suffix
            counter = 1
            while (destination_folder / f"{base}_{counter}{ext}").exists():
                counter += 1
            dest_path = destination_folder / f"{base}_{counter}{ext}"

        shutil.move(str(src_path), str(dest_path))
        print(f"✅ Moved {src_path.name} → {dest_path}")

    print("🎉 Done.")

if __name__ == "__main__":
    # default destination_relative is "../folder_2" (one level up into folder_2)
    # you can change it, e.g. move_json_files(destination_relative="../../data/folder2")
    move_json_files()
```
